```python
from pathlib import Path
import fastf1
import pandas as pd


# Create the cache directory if it doesn't exist
cache_dir = Path("cache")
cache_dir.mkdir(parents=True, exist_ok=True)

# Enable FastF1 cache
fastf1.Cache.enable_cache(str(cache_dir))
```

```python
GP2024 = ["Bahrain", "Saudi Arabia", "Australia", "Japan", "China",
          "Miami", "Emilia-Romagna", "Monaco", "Canada", "Spain",
          "Austria", "Great Britain", "Hungary", "Belgium", "Netherlands",
          "Italy", "Azerbaijan", "Singapore", "United States", "Mexico",
          "Brazil", "Las Vegas", "Qatar", "Abu Dhabi"]

def get_session(gp, year=2026):
    if gp not in GP2024:
        raise ValueError(f"Grand Prix '{gp}' not found in GP2024.")

    session = fastf1.get_session(year, gp, "R")

    print(f"Loading {year} {gp} Grand Prix race session...")

    session.load(
        laps=True,
        telemetry=False,
        weather=False,
        messages=False
    )

    print("Session loading completed.")
    print("Laps available:", hasattr(session, "_laps"))

    return session

session = get_session("Monaco", 2026)
```
